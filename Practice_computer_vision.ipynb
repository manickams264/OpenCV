```python
img = cv2.imread("/content/image_opencv.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
edges = cv2.Canny(gray, 50, 150)
lines = cv2.HoughLines(edges, 1, np.pi/180, 100)
for line in lines:
    rho, theta = line[0]
    a = np.cos(theta)
    b = np.sin(theta)
    x0 = a * rho
    y0 = b * rho
    x1 = int(x0 + 1000 * (-b))
    y1 = int(y0 + 1000 * (a))
    x2 = int(x0 - 1000 * (-b))
    y2 = int(y0 - 1000 * (a))
    cv2.line(img, (x1, y1), (x2, y2), (0, 0, 255), 2)
from google.colab.patches import cv2_imshow
cv2_imshow(img)
```

```python
img = cv2.imread("/content/image_opencv.jpg")
gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
circles = cv2.HoughCircles(gray_img, cv2.HOUGH_GRADIENT, 1, 20)
if circles is not None:
    circles = np.uint16(np.around(circles))
    for c in circles[0, :]:
        cv2.circle(img, (c[0], c[1]), c[2], (0, 255, 0), 2)
from google.colab.patches import cv2_imshow
cv2_imshow(img)
```

```python
img = cv2.imread("/content/image_opencv.jpg")
h, w = img.shape[:2]
map_x = np.zeros((h, w), np.float32)
map_y = np.zeros((h, w), np.float32)
for i in range(h):
    for j in range(w):
        map_x[i, j] = w - j - 1
        map_y[i, j] = i
remap = cv2.remap(img, map_x, map_y, cv2.INTER_LINEAR)
from google.colab.patches import cv2_imshow
cv2_imshow(remap)
```
